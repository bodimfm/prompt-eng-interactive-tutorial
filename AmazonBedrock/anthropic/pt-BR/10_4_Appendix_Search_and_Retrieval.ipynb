# Appendix 10.3: Search & Retrieval

Você sabia que pode usar Claude para **pesquisar na Wikipedia para você**? Claude pode encontrar e recuperar artigos, momento em que você também pode usar Claude para resumir e sintetizá-los, escrever conteúdo novo a partir do que encontrou, e muito mais. E não apenas Wikipedia! Você também pode pesquisar em seus próprios documentos, sejam armazenados como texto simples ou incorporados em um banco de dados vetorial.

Veja nossos [exemplos de cookbook RAG](https://github.com/anthropics/anthropic-cookbook/blob/main/third_party/Wikipedia/wikipedia-search-cookbook.ipynb) para aprender como complementar o conhecimento do Claude e melhorar a precisão e relevância das respostas do Claude com dados recuperados de bancos de dados vetoriais, Wikipedia, internet e muito mais. Lá, você também pode aprender sobre como usar certos [embeddings](https://docs.anthropic.com/claude/docs/embeddings) e ferramentas de banco de dados vetorial.

Se você está interessado em aprender sobre arquiteturas RAG avançadas usando Claude, confira nossos [slides de apresentação técnica do Claude 3 sobre arquiteturas RAG](https://docs.google.com/presentation/d/1zxkSI7lLUBrZycA-_znwqu8DDyVhHLkQGScvzaZrUns/edit#slide=id.g2c736259dac_63_782).

### O SDK da Anthropic e a API Messages
Usaremos o [SDK Python da Anthropic](https://docs.anthropic.com/claude/reference/claude-on-amazon-bedrock) e a [API Messages](https://docs.anthropic.com/claude/reference/messages_post) durante todo este tutorial.

Abaixo está um exemplo de como será a execução de um prompt neste tutorial.

Primeiro, definimos e armazenamos o nome do modelo e a região.

```python
import boto3
session = boto3.Session() # create a boto3 session to dynamically get and set the region name
AWS_REGION = session.region_name
print("AWS Region:", AWS_REGION)
MODEL_NAME = "anthropic.claude-3-haiku-20240307-v1:0"

%store MODEL_NAME
%store AWS_REGION
```

Em seguida, criamos `get_completion`, que é uma função auxiliar que envia um prompt para Claude e retorna a resposta gerada por Claude. Execute essa célula agora.

```python
from anthropic import AnthropicBedrock

client = AnthropicBedrock(aws_region=AWS_REGION)

def get_completion(prompt, system=''):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=2000,
        temperature=0.0,
        messages=[
          {"role": "user", "content": prompt}
        ],
        system=system
    )
    return message.content[0].text
```

Agora vamos escrever um exemplo de prompt para Claude e imprimir a saída de Claude executando nossa função auxiliar `get_completion`. A execução da célula abaixo imprimirá uma resposta de Claude abaixo dela.

Sinta-se à vontade para experimentar com a string do prompt para obter respostas diferentes de Claude.

```python
# Prompt
prompt = "Hello, Claude!"

# Get Claude's response
print(get_completion(prompt))
```

As variáveis `MODEL_NAME` e `AWS_REGION` definidas anteriormente serão usadas durante todo o tutorial. Apenas certifique-se de executar as células de cada página do tutorial de cima para baixo.

## Lição

Continuando no tema de que o Claude não tem contexto além do que você diz, às vezes é importante **fazer o Claude assumir um papel específico (incluindo todo o contexto necessário)**. Isso também é conhecido como role prompting. Quanto mais detalhes no contexto do papel, melhor.

**Preparar o Claude com um papel pode melhorar o desempenho do Claude** em uma variedade de áreas, desde escrita até codificação e resumos. É como quando humanos podem às vezes ser ajudados quando são instruídos a "pensar como um ______". Role prompting também pode mudar o estilo, tom e maneira da resposta do Claude.

**Nota:** Role prompting pode acontecer tanto no system prompt quanto como parte do turno de mensagem do User.

### Exemplos

No exemplo abaixo, vemos que sem role prompting, o Claude fornece uma **resposta direta e sem estilo** quando perguntado para dar uma perspectiva de uma única frase sobre skate.

No entanto, quando preparamos o Claude para assumir o papel de um gato, a perspectiva do Claude muda, e assim **o tom, estilo e conteúdo da resposta do Claude se adaptam ao novo papel**.

**Nota:** Uma técnica bônus que você pode usar é **fornecer ao Claude contexto sobre seu público-alvo**. Abaixo, poderíamos ter ajustado o prompt para também dizer ao Claude com quem ele deveria estar falando. "Você é um gato" produz uma resposta bastante diferente de "você é um gato falando para uma multidão de skatistas."

Aqui está o prompt sem role prompting no system prompt:

```python
# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Print Claude's response
print(get_completion(PROMPT))
```

Aqui está a mesma pergunta do usuário, exceto com role prompting.

```python
# System prompt
SYSTEM_PROMPT = "You are a cat."

# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Print Claude's response
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

Você pode usar role prompting como uma forma de fazer o Claude emular certos estilos na escrita, falar com uma certa voz, ou guiar a complexidade de suas respostas. **Role prompting também pode tornar o Claude melhor em realizar tarefas de matemática ou lógica.**

Por exemplo, no exemplo abaixo, há uma resposta definitivamente correta, que é sim. No entanto, o Claude erra e pensa que falta informação, mas não falta:

```python
# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don’t know if Anne is married. Is a married person looking at an unmarried person?"

# Print Claude's response
print(get_completion(PROMPT))
```

Agora, e se **prepararmos o Claude para agir como um bot de lógica**? Como isso mudará a resposta do Claude?

Descobre-se que com esta nova atribuição de papel, o Claude acerta. (Embora notavelmente não por todas as razões certas)

```python
# System prompt
SYSTEM_PROMPT = "You are a logic bot designed to answer complex logic problems."

# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don’t know if Anne is married. Is a married person looking at an unmarried person?"

# Print Claude's response
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

**Nota:** O que você aprenderá ao longo deste curso é que existem **muitas técnicas de engenharia de prompt que você pode usar para obter resultados similares**. Quais técnicas você usa depende de você e sua preferência! Encorajamos você a **experimentar para encontrar seu próprio estilo de engenharia de prompt**.

### Exercício 3.1 - Correção de Matemática
Em algumas instâncias, **Claude pode ter dificuldades com matemática**, mesmo matemática simples. Abaixo, o Claude avalia incorretamente o problema de matemática como corretamente resolvido, embora haja um erro aritmético óbvio no segundo passo. Note que o Claude na verdade detecta o erro quando passa passo a passo, mas não chega à conclusão de que a solução geral está errada.

Modifique o `PROMPT` e/ou o `SYSTEM_PROMPT` para fazer o Claude avaliar a solução como resolvida `incorretamente`, em vez de corretamente resolvida.

```python
# System prompt - if you don't want to use a system prompt, you can leave this variable set to an empty string
SYSTEM_PROMPT = ""

# Prompt
PROMPT = """Is this equation solved correctly below?

2x - 3 = 9
2x = 6
x = 3"""

# Get Claude's response
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    if "incorrect" in text or "not correct" in text.lower():
        return True
    else:
        return False

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

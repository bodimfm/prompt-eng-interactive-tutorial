## Lição

Muitas vezes, não queremos escrever prompts completos, mas sim queremos **templates de prompt que podem ser modificados mais tarde com dados de entrada adicionais antes de enviar ao Claude**. Isso pode ser útil se você quer que o Claude faça a mesma coisa todas as vezes, mas os dados que o Claude usa para sua tarefa podem ser diferentes a cada vez.

Felizmente, podemos fazer isso facilmente **separando o esqueleto fixo do prompt da entrada de usuário variável, e então substituindo a entrada de usuário no prompt** antes de enviar o prompt completo ao Claude.

Abaixo, vamos percorrer passo a passo como escrever um template de prompt substituível, bem como como substituir a entrada do usuário.

### Exemplos

Neste primeiro exemplo, estamos pedindo ao Claude para agir como um gerador de sons de animais. Note que o prompt completo enviado ao Claude é apenas o `PROMPT_TEMPLATE` substituído com a entrada (neste caso, "Cow"). Note que a palavra "Cow" substitui o placeholder `ANIMAL` via uma f-string quando imprimimos o prompt completo.

**Nota:** Você não precisa chamar sua variável placeholder de algo em particular na prática. Nós a chamamos de `ANIMAL` neste exemplo, mas facilmente poderíamos tê-la chamado de `CREATURE` ou `A` (embora seja geralmente bom ter seus nomes de variáveis específicos e relevantes para que seu template de prompt seja fácil de entender mesmo sem a substituição, apenas para parseabilidade do usuário). Apenas certifique-se de que seja qual for o nome que você dê à sua variável, é o que você usa para a f-string do template de prompt.

```python
# Variable content
ANIMAL = "Cow"

# Prompt template with a placeholder for the variable content
PROMPT = f"I will tell you the name of an animal. Please respond with the noise that animal makes. {ANIMAL}"

# Print Claude's response
print("--------------------------- Full prompt with variable substutions ---------------------------")
print(PROMPT)
print("\n------------------------------------- Claude's response -------------------------------------")
print(get_completion(PROMPT))
```

Por que gostaríamos de separar e substituir entradas assim? Bem, **templates de prompt simplificam tarefas repetitivas**. Digamos que você construa uma estrutura de prompt que convida usuários terceiros a submeter conteúdo ao prompt (neste caso o animal cujo som eles querem gerar). Esses usuários terceiros não precisam escrever ou mesmo ver o prompt completo. Tudo o que eles têm que fazer é preencher variáveis.

Fazemos essa substituição aqui usando variáveis e f-strings, mas você também pode fazer isso com o método format().

**Nota:** Templates de prompt podem ter quantas variáveis você desejar!

Ao introduzir variáveis de substituição assim, é muito importante **garantir que o Claude saiba onde as variáveis começam e terminam** (vs. instruções ou descrições de tarefas). Vamos ver um exemplo onde não há separação entre as instruções e a variável de substituição.

Para nossos olhos humanos, é muito claro onde a variável começa e termina no template de prompt abaixo. No entanto, no prompt totalmente substituído, essa delimitação se torna pouco clara.

```python
# Variable content
EMAIL = "Show up at 6am tomorrow because I'm the CEO and I say so."

# Prompt template with a placeholder for the variable content
PROMPT = f"Yo Claude. {EMAIL} <----- Make this email more polite but don't change anything else about it."

# Print Claude's response
print("--------------------------- Full prompt with variable substutions ---------------------------")
print(PROMPT)
print("\n------------------------------------- Claude's response -------------------------------------")
print(get_completion(PROMPT))
```

Aqui, **o Claude acha que "Yo Claude" é parte do email que ele deveria reescrever**! Você pode dizer isso porque ele começa sua reescrita com "Dear Claude". Para o olho humano, é claro, particularmente no template de prompt onde o email começa e termina, mas se torna muito menos claro no prompt após a substituição.

Como resolvemos isso? **Envolva a entrada em tags XML**! Fizemos isso abaixo, e como você pode ver, não há mais "Dear Claude" na saída.

[Tags XML](https://docs.anthropic.com/claude/docs/use-xml-tags) são tags com colchetes angulares como `<tag></tag>`. Elas vêm em pares e consistem em uma tag de abertura, como `<tag>`, e uma tag de fechamento marcada por uma `/`, como `</tag>`. Tags XML são usadas para envolver conteúdo, assim: `<tag>content</tag>`.

**Nota:** Embora o Claude possa reconhecer e trabalhar com uma ampla gama de separadores e delimitadores, recomendamos que você **use especificamente tags XML como separadores** para o Claude, já que o Claude foi treinado especificamente para reconhecer tags XML como um mecanismo de organização de prompt. Fora da chamada de função, **não há tags XML de molho especial nas quais o Claude foi treinado que você deva usar para maximizar seu desempenho**. Fizemos o Claude intencionalmente muito maleável e customizável dessa forma.

```python
# Variable content
EMAIL = "Show up at 6am tomorrow because I'm the CEO and I say so."

# Prompt template with a placeholder for the variable content
PROMPT = f"Yo Claude. <email>{EMAIL}</email> <----- Make this email more polite but don't change anything else about it."

# Print Claude's response
print("--------------------------- Full prompt with variable substutions ---------------------------")
print(PROMPT)
print("\n------------------------------------- Claude's response -------------------------------------")
print(get_completion(PROMPT))
```

Vamos ver outro exemplo de como tags XML podem nos ajudar.

No prompt a seguir, **o Claude interpreta incorretamente qual parte do prompt é a instrução vs. a entrada**. Ele considera incorretamente `Each is about an animal, like rabbits` como parte da lista devido à formatação, quando o usuário (aquele preenchendo a variável `SENTENCES`) presumivelmente não queria isso.

```python
# Variable content
SENTENCES = """- I like how cows sound
- This sentence is about spiders
- This sentence may appear to be about dogs but it's actually about pigs"""

# Prompt template with a placeholder for the variable content
PROMPT = f"""Below is a list of sentences. Tell me the second item on the list.

- Each is about an animal, like rabbits.
{SENTENCES}"""

# Print Claude's response
print("--------------------------- Full prompt with variable substutions ---------------------------")
print(PROMPT)
print("\n------------------------------------- Claude's response -------------------------------------")
print(get_completion(PROMPT))
```

Para corrigir isso, só precisamos **envolver as frases de entrada do usuário em tags XML**. Isso mostra ao Claude onde os dados de entrada começam e terminam, apesar do hífen enganoso antes de `Each is about an animal, like rabbits.`

```python
# Variable content
SENTENCES = """- I like how cows sound
- This sentence is about spiders
- This sentence may appear to be about dogs but it's actually about pigs"""

# Prompt template with a placeholder for the variable content
PROMPT = f""" Below is a list of sentences. Tell me the second item on the list.

- Each is about an animal, like rabbits.
<sentences>
{SENTENCES}
</sentences>"""

# Print Claude's response
print("--------------------------- Full prompt with variable substutions ---------------------------")
print(PROMPT)
print("\n------------------------------------- Claude's response -------------------------------------")
print(get_completion(PROMPT))
```

**Nota:** Na versão incorreta do prompt "Each is about an animal", tivemos que incluir o hífen para fazer o Claude responder incorretamente da forma que queríamos para este exemplo. Esta é uma lição importante sobre prompting: **pequenos detalhes importam**! Sempre vale a pena **revisar seus prompts para erros de digitação e erros gramaticais**. O Claude é sensível a padrões (em seus primeiros anos, antes do fine-tuning, era uma ferramenta bruta de previsão de texto), e é mais provável que cometa erros quando você comete erros, mais inteligente quando você parece inteligente, mais bobo quando você parece bobo, e assim por diante.

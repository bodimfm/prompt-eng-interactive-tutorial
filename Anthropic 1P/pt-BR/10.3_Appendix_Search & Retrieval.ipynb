# Apêndice 10.3: Busca e Recuperação

Você sabia que pode usar o Claude para **buscar na Wikipedia para você**? O Claude pode encontrar e recuperar artigos, momento em que você também pode usar o Claude para resumir e sintetizá-los, escrever conteúdo novo a partir do que encontrou, e muito mais. E não apenas Wikipedia! Você também pode buscar em seus próprios documentos, sejam armazenados como texto simples ou incorporados em um armazenamento de dados vetoriais.

Veja nossos [exemplos de cookbook RAG](https://github.com/anthropics/anthropic-cookbook/blob/main/third_party/Wikipedia/wikipedia-search-cookbook.ipynb) para aprender como complementar o conhecimento do Claude e melhorar a precisão e relevância das respostas do Claude com dados recuperados de bancos de dados vetoriais, Wikipedia, internet e muito mais. Lá, você também pode aprender sobre como usar certas ferramentas de [embeddings](https://docs.anthropic.com/claude/docs/embeddings) e banco de dados vetoriais.

Se você está interessado em aprender sobre arquiteturas RAG avançadas usando o Claude, confira nossos [slides de apresentação técnica do Claude 3 sobre arquiteturas RAG](https://docs.google.com/presentation/d/1zxkSI7lLUBrZycA-_znwqu8DDyVhHLkQGScvzaZrUns/edit#slide=id.g2c736259dac_63_782).

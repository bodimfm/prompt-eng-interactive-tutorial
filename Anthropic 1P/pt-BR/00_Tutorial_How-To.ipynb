### O SDK da Anthropic e a Messages API
Usaremos o [SDK Python da Anthropic](https://docs.anthropic.com/claude/reference/client-sdks) e a [Messages API](https://docs.anthropic.com/claude/reference/messages_post) ao longo deste tutorial.

Abaixo está um exemplo de como será a execução de um prompt neste tutorial. Primeiro, criamos `get_completion`, que é uma função auxiliar que envia um prompt para o Claude e retorna a resposta gerada pelo Claude. Execute essa célula agora.

```python
import anthropic

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=2000,
        temperature=0.0,
        messages=[
          {"role": "user", "content": prompt}
        ]
    )
    return message.content[0].text
```

Agora vamos escrever um exemplo de prompt para o Claude e imprimir a saída do Claude executando nossa função auxiliar `get_completion`. Executar a célula abaixo imprimirá uma resposta do Claude abaixo dela.

Sinta-se à vontade para experimentar com a string do prompt para obter diferentes respostas do Claude.

```python
# Prompt
prompt = "Hello, Claude!"

# Get Claude's response
print(get_completion(prompt))
```

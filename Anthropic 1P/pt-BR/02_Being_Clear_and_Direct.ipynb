## Lição

**Claude responde melhor a instruções claras e diretas.**

Pense no Claude como qualquer outro humano que é novo no trabalho. **Claude não tem contexto** sobre o que fazer além do que você literalmente diz a ele. Assim como quando você instrui um humano pela primeira vez em uma tarefa, quanto mais você explica exatamente o que quer de maneira direta ao Claude, melhor e mais precisa será a resposta do Claude.

Quando estiver em dúvida, siga a **Regra de Ouro de Prompts Claros**:
- Mostre seu prompt para um colega ou amigo e peça que eles sigam as instruções para ver se conseguem produzir o resultado que você deseja. Se eles ficarem confusos, o Claude ficará confuso.

### Exemplos

Vamos pegar uma tarefa como escrever poesia. (Ignore qualquer incompatibilidade de sílabas - LLMs ainda não são ótimos em contar sílabas.)

```python
# Prompt
PROMPT = "Write a haiku about robots."

# Print Claude's response
print(get_completion(PROMPT))
```

Este haiku é bom o suficiente, mas os usuários podem querer que o Claude vá direto ao poema sem o preâmbulo "Here is a haiku".

Como conseguimos isso? Nós **pedimos por isso**!

```python
# Prompt
PROMPT = "Write a haiku about robots. Skip the preamble; go straight into the poem."

# Print Claude's response
print(get_completion(PROMPT))
```

Aqui está outro exemplo. Vamos perguntar ao Claude quem é o melhor jogador de basquete de todos os tempos. Você pode ver abaixo que, embora o Claude liste alguns nomes, **ele não responde com um "melhor" definitivo**.

```python
# Prompt
PROMPT = "Who is the best basketball player of all time?"

# Print Claude's response
print(get_completion(PROMPT))
```

Podemos fazer o Claude se decidir e escolher um melhor jogador? Sim! Apenas peça!

```python
# Prompt
PROMPT = "Who is the best basketball player of all time? Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"

# Print Claude's response
print(get_completion(PROMPT))
```

### Exercício 2.1 - Espanhol
Modifique o `SYSTEM_PROMPT` para fazer o Claude gerar sua resposta em espanhol.

```python
# System prompt - this is the only field you should chnage
SYSTEM_PROMPT = "[Replace this text]"

# Prompt
PROMPT = "Hello Claude, how are you?"

# Get Claude's response
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    return "hola" in text.lower()

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

Modifique o `PROMPT` para que o Claude não hesite de forma alguma e responda com **APENAS** o nome de um jogador específico, sem **nenhuma outra palavra ou pontuação**.

### Exercício 2.3 - Escreva uma História

Modifique o `PROMPT` para que o Claude responda com uma resposta tão longa quanto você conseguir. Se sua resposta tiver **mais de 800 palavras**, a resposta do Claude será avaliada como correta.

```python
# Prompt - this is the only field you should change
PROMPT = "[Replace this text]"

# Get Claude's response
response = get_completion(PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    trimmed = text.strip()
    words = len(trimmed.split())
    return words >= 800

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```
